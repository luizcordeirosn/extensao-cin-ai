# 🧠 Desafio 01 - Representação de Dados

### 📝 Estratégia Adotada

Inicialmente, considerei utilizar a biblioteca `scikit-learn` com o `TfidfVectorizer` para transformar o texto em uma representação numérica, baseado em um projeto anterior que trabalhei durante minha graduação. No entanto, como essa abordagem atribui **pesos proporcionais à frequência das palavras**, não garantiria a possibilidade de **reconstruir o texto original**.

Diante disso, optei por uma abordagem mais direta: representar o texto por meio dos **códigos Unicode de cada caractere**, utilizando a função `ord()`. Essa conversão é **totalmente reversível** com a função `chr()`, o que permite transformar a matriz numérica de volta ao texto original sem perda de informação (exceto pontuações e acentuação, que foram removidas previamente).

---

### 🔧 Etapas do processo

1. **Leitura do texto** a partir do arquivo `Base Text - Matrix Representation I.txt` hospedado no Google Drive.
2. **Limpeza do conteúdo**, removendo pontuações e caracteres especiais para manter apenas letras e espaços.
3. **Separação por linhas** (`split('\n')`) e iteração linha a linha.
4. Para cada caractere das linhas, aplicou-se `ord(c)` para obter o valor numérico correspondente.
5. As listas numéricas foram organizadas em uma **matriz de inteiros** (lista de listas).
6. Para a **reconstrução do texto**, aplicou-se a função `chr(n)` em cada número da matriz, reagrupando os caracteres por linha com `'\n'.join(...)`.

---

Essa abordagem garante uma representação numérica **consistente e totalmente reversível** do texto original, atendendo aos critérios definidos no desafio.



```python
# imports
from google.colab import drive
import re
```

```python
# get file from drive
def get_file_from_drive():

  drive.mount('/content/drive')

  path = '/content/drive/MyDrive/Base Text - Matrix Representation I.txt'

  with open(path, 'r') as f:
    return f.read()

# word processing
def word_processing(content):

  return re.sub(r'[^a-zA-Z0-9\s]', '', content)

# enumerating a numeric array from initial text
def initial_text_to_numeric_array(content):
  rows = content.split('\n')

  return [[ord(word) for word in row] for row in rows]

# converting a numeric array to original text
def numeric_array_to_original_text(numeric_array):

  return '\n'.join(''.join(chr(number) for number in row) for row in numeric_array)
```

```python
content = get_file_from_drive()
content_after_word_processing = word_processing(content)
numeric_array = initial_text_to_numeric_array(content_after_word_processing)
original_text = numeric_array_to_original_text(numeric_array)

print()
print('Matriz numérica: ')
print(numeric_array)

print()
print('Conversão da matriz numérica para o texto original: ')
print(original_text)
```
